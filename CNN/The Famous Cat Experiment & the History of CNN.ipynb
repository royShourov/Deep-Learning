#### **CNN vs Visual Cortex: The Famous Cat Experiment & the History of CNN**

#### **The Big Idea**

The human (and cat) brain recognizes objects through hierarchical feature detection:

* First detect edges
* Then combine edges into shapes
* Then combine shapes into objects

Modern CNNs follow almost the same philosophy. This is why CNNs are often called brain-inspired neural networks, although they are not exact copies of the biological visual system.

#### **Before CNNs: The Problem**

In the 1950s, scientists had a huge question:

> How does the brain recognize objects?

When we look at a face, we instantly recognize it. But how?

* Does one neuron recognize a whole face?
* Or do many neurons work together?

No one knew. Two neuroscientists—David Hubel and Torsten Wiesel—decided to find the answer. Their discoveries later earned them the 1981 Nobel Prize.

![Understanding CNN for Image Processing | Svitla Systems](https://images.openai.com/static-rsc-4/rEYAZZHlvOHXcMgpPcs7Vtb5HRlaHPv53x6EmLN7MYLtE09JZ7h9VgLSRCS35KaOOQxPPLgw-Dcniu1zlbl1uCDtPqMGb84-gflzLV4SMDXbEZ1cgbR25AtTTks7LL0auc4nqksE5zF_B1YKC3INixJdTEc6GBksMhfCKouQYhY?purpose=inline)

This experiment became one of the most famous experiments in neuroscience.

#### **Experimental Setup**

The researchers:

1. Used an anesthetized cat.
2. Inserted a tiny microelectrode into the cat's Primary Visual Cortex (V1).
3. Displayed different visual patterns on a screen.

The electrode recorded electrical activity from a single neuron.

#### **An Unexpected Discovery**

Initially, they projected various shapes. Nothing exciting happened. Then something accidental occurred. While changing slides, a thin shadow line moved across the screen. Suddenly—

> The neuron fired rapidly.

That neuron wasn't responding to the whole picture. It loved a line at a particular angle. This accidental observation changed neuroscience forever.

#### **What They Discovered**

They found two important kinds of neurons.

#### **1. Simple Cells**

![6.5: The Thalamus and Primary Visual Cortex - Social Sci LibreTexts](https://images.openai.com/static-rsc-4/okmREN0X4-72Ak2IlNg5ZCYod-tILHyOuauJj4XgHz-bG2uB6T0CUNXaAEo0P_hS3OBROFepUIiBMplvZScfzfv7wbUmMqMnn9iZZqdsolyvve7sAPIhfVit0YZ0xQ4rIeEdWIRf8UJZjWYlia-20sLCi8FyO9gk9Z8NrzcDIo4?purpose=inline)

Simple cells respond to:

* Vertical edges
* Horizontal edges
* Diagonal edges

But only at a specific location with a specific orientation. Example:

| Stimulus | Response |
| :--- | :--- |
| Vertical line | Strong |
| Horizontal line | Weak |
| Diagonal line | Medium |

#### **2. Complex Cells**

Complex cells behave differently. Instead of caring about one exact location, they recognize the same edge even if it moves slightly. Think of this:

```
|      → detected
  |    → detected
    |  → detected
```

The position changes. The neuron still recognizes it. This idea became incredibly important for CNNs.

# How the Visual Cortex Processes Images

The brain processes information layer by layer.

![深度学习简介(一)——卷积神经网络 - alexcai - 博客园](https://images.openai.com/static-rsc-4/-pgwd9qYLqv3S9qRlOqm4D2UHEzlYyyhmA5i8AoDAmjsIGNeYeIjw1H4HrdQnvMBFF5vTYk1XQ6LRYDrwbERIGsJYvqt2hTw3il2rTxuOPi971YfipRiOs72edPFNv8RWZU0FbNEJDO6OyUamRScWAuGtdChOyzrVyEWXq7uWM4?purpose=inline)

| Brain Layer | Detects |
| :--- | :--- |
| Early | Edges |
| Middle | Corners |
| Higher | Shapes |
| Highest | Objects |

This hierarchy inspired CNN architecture.

#### **Receptive Fields**

One of Hubel and Wiesel's biggest discoveries was the receptive field. A receptive field is:

> the small region of an image that affects one neuron.

Instead of looking at the whole image, each neuron looks at only a tiny patch. 
Example:

#### **The Birth of CNN**

After Hubel and Wiesel's discoveries computer scientists wondered:

> Can we build an artificial vision system using these ideas?

The answer came from Kunihiko Fukushima.

![Historias de la IA: el cognitrón de Kunihiko Fukushima - Matemáticas y sus fronteras](https://images.openai.com/static-rsc-4/h5YyY1cpAD0_g4LK15tpwD7LsSlWrMQ1qFi7XHLohHL5BJBtPqi1o1qKbjwM1UXQjUOhIN8_U22-rn1QgTD_Yb3L9zbwx4PII2ov0vFoLVYiPrS1wsf8di2kefdt4wHXDIX_XUbcEqc78zcS8KPJHixHCyw7es2_mK50x-oG684?purpose=inline)


#### **1969–1980: The Neocognitron**

Fukushima designed the Neocognitron. It introduced ideas that resemble today's CNN:

* Local receptive fields
* Shared filters
* Hierarchical processing
* Position tolerance

However, its weights were not trained with modern backpropagation. It learned differently. Still, it became the direct ancestor of CNNs.

#### **2012: AlexNet Changed Everything**

AlexNet achieved a massive improvement on the ImageNet competition. Why?

* GPUs
* ReLU
* Dropout
* Deep architecture

This success launched the deep learning revolution.

#### **Important Differences**

Although CNNs were inspired by the brain, they are not biologically identical.

| Brain | CNN |
| :--- | :--- |
| Millions of neuron types | Uniform artificial neurons |
| Feedback connections | Mostly feedforward |
| Learns continuously | Usually trained once |
| Extremely energy efficient | Requires large computation |

Modern neuroscience now shows the brain uses many recurrent and feedback pathways that traditional CNNs do not fully capture.

#### **Why the Cat Experiment Still Matters**

The Hubel–Wiesel experiment introduced three revolutionary ideas:

1. Receptive fields – neurons look at small regions.
2. Orientation selectivity – neurons detect specific edges.
3. Hierarchical processing – simple features combine into complex objects.

These three ideas became the conceptual foundation of CNNs.Without that cat experiment, modern computer vision—including face recognition, self-driving cars, medical imaging and image search—would likely have developed very differently.
